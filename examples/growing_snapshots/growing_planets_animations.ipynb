```python
solar_system_df = pd.read_csv('solarsystem.csv')
mercury = solar_system_df[solar_system_df['name'] == 'Mercury']
venus = solar_system_df[solar_system_df['name'] == 'Venus']
earth = solar_system_df[solar_system_df['name'] == 'Earth']
mars = solar_system_df[solar_system_df['name'] == 'Mars']
```

```python
base_path = "/Users/honeyeah/Codes/data/" #folder with all data


sim_set = "cham_bi_tajer/" #name of simulation set, folder with subfolders the name "sim_set_i"
sim_archive_path = base_path + sim_set + "cham_bi_tajer_" #path given to get_runtime_data function
cham_bi_tajer_rdf = rr.get_runtime_data(sim_archive_path, 50, "cham_simarchive_") #rdf is runtime data frame


cut_time = 299.9e6
```

```python
n_runs = 50

maxtimes_cham_bi_tajer = np.array([
    cham_bi_tajer_rdf[cham_bi_tajer_rdf['simulation_id'] == i + 1]['time'].max()
    for i in range(n_runs)
])

keep_cham_bi_tajer = np.where(maxtimes_cham_bi_tajer > 299.9e6)[0]

print("number of finished runs\n")
print(f"cham_bi_tajer: {len(keep_cham_bi_tajer)}")
```

```python
#FOR cham_bi_tajer
cut_times = np.logspace(6, np.log10(299) + 6, 20)
cham_bi_tajer_u_dataframes = []

for cut in cut_times:
    cham_bi_tajer_ppdf_list = rr.get_particle_params_at_time(
        f"/Users/honeyeah/Codes/data/cham_bi_tajer/cham_bi_tajer_", keep_cham_bi_tajer, "cham_simarchive_", cut
    )

    cham_bi_tajer_u_pldf_list = []
    for j, i in enumerate(keep_cham_bi_tajer):
        df = rr.read_dbct_output(
        f"/Users/honeyeah/Codes/data/cham_bi_tajer/cham_bi_tajer_{i+1}/animation/uniform_33_{int(cut/1e6)}Myrs.txt",
        0.3, "none", "none"
        )

        df_len = len(df)  
        param_len = len(cham_bi_tajer_ppdf_list[j])
        min_len = min(df_len, param_len)

        if min_len == 0:
            print(f"Skipping index {i}: empty df or param list.")
            continue

        # Trim both to same length
        df = df.iloc[:min_len].copy()
        semi_values = cham_bi_tajer_ppdf_list[j]['semi'][:min_len]
        df['semi'] = semi_values.values  # if semi_values is a Series
        df['sim_id'] = i + 1

        cham_bi_tajer_u_pldf_list.append(df)
        df_at_snapshot = pd.concat(cham_bi_tajer_u_pldf_list)
        df_at_snapshot['time'] = cut
        cham_bi_tajer_u_dataframes.append(df_at_snapshot)

cham_bi_tajer_u_pdframe = pd.concat(cham_bi_tajer_u_dataframes)
```

```python
#Step function
cham_bi_tajer_s_dataframes = []

for cut in cut_times:
    cham_bi_tajer_ppdf_list = rr.get_particle_params_at_time(
        f"/Users/honeyeah/Codes/data/cham_bi_tajer/cham_bi_tajer_", keep_cham_bi_tajer, "cham_simarchive_", cut
    )

    cham_bi_tajer_s_pldf_list = []
    for j, i in enumerate(keep_cham_bi_tajer):
        df = rr.read_dbct_output(
        f"/Users/honeyeah/Codes/data/cham_bi_tajer/cham_bi_tajer_{i+1}/animation/step_b0.7_p0.20_{int(cut/1e6)}Myrs.txt",
        0.3, "none", "none"
        )

        df_len = len(df)  
        param_len = len(cham_bi_tajer_ppdf_list[j])
        min_len = min(df_len, param_len)

        if min_len == 0:
            print(f"Skipping index {i}: empty df or param list.")
            continue

        # Trim both to same length
        df = df.iloc[:min_len].copy()
        semi_values = cham_bi_tajer_ppdf_list[j]['semi'][:min_len]
        df['semi'] = semi_values.values  # if semi_values is a Series
        df['sim_id'] = i + 1

        cham_bi_tajer_s_pldf_list.append(df)
        df_at_snapshot = pd.concat(cham_bi_tajer_s_pldf_list)
        df_at_snapshot['time'] = cut
        cham_bi_tajer_s_dataframes.append(df_at_snapshot)

cham_bi_tajer_s_pdframe = pd.concat(cham_bi_tajer_s_dataframes)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

dot_scale = 600
col_labels = ["Uniform", "Step function"]

fig, axes = plt.subplots(1, 2, figsize=(10, 5), sharey=True, sharex=True)

# set up the axes
for col, ax in enumerate(axes):
    ax.set_xlim(0.2, 2.2)
    ax.set_ylim(-0.1, 1.1)
    ax.grid(True, alpha=0.5)
    ax.plot([0.2, 2.2], [0.33, 0.33], linestyle='dashed', color='grey', alpha=0.5)
    ax.set_title(col_labels[col])

fig.text(0.5, 0.01, 'Semi-major axis (AU)', ha='center')
fig.text(0.05, 0.5, r'CMF', va='center', rotation='vertical')

# scatter artists (placeholders, will be updated)
scatters = [axes[0].scatter([], [], s=[]), axes[1].scatter([], [], s=[])]

def init():
    for sc in scatters:
        sc.set_offsets(np.empty((0, 2)))
        sc.set_sizes([])
    fig.suptitle("")
    return scatters

def update(frame):
    t = cut_times[frame]

    # Select rows for this time
    df_u = cham_bi_tajer_u_pdframe[cham_bi_tajer_u_pdframe['time'] == t]
    df_s = cham_bi_tajer_s_pdframe[cham_bi_tajer_s_pdframe['time'] == t]

    # Update Uniform scatter
    offsets_u = np.column_stack((df_u['semi'], df_u['cmf']))
    scatters[0].set_offsets(offsets_u)
    scatters[0].set_sizes(df_u['mass'] * sol_to_earth * dot_scale)
    scatters[0].set_facecolors("grey")
    scatters[0].set_edgecolors("white")
    scatters[0].set_alpha(0.4)

    # Update Step scatter
    offsets_s = np.column_stack((df_s['semi'], df_s['cmf']))
    scatters[1].set_offsets(offsets_s)
    scatters[1].set_sizes(df_s['mass'] * sol_to_earth * dot_scale)
    scatters[1].set_facecolors("grey")
    scatters[1].set_edgecolors("white")
    scatters[1].set_alpha(0.4)

    # update title
    fig.suptitle(f"t = {t/1e6:.2f} Myrs")

     # --- Add Solar System planets (fixed reference points) ---
    for ax in axes:
        ax.scatter(0.387, 0.7, color='#D81B60', edgecolors='black', linewidths=1,
                   s=0.055 * dot_scale, label='Mercury')
        ax.annotate('☿', [0.4, 0.71], fontsize=12, usetex=False)

        ax.scatter(0.723, 0.3, color='#FFC107', edgecolors='black', linewidths=1,
                   s=0.815 * dot_scale, label='Venus')
        ax.annotate('♀', [0.68, 0.2], fontsize=12, usetex=False)

        ax.scatter(1, 0.33, color='#1E88E5', edgecolors='black', linewidths=1,
                   s=dot_scale, label='Earth')
        ax.annotate('⊕', [1.03, 0.38], fontsize=12, usetex=False)

        ax.scatter(1.52, 0.24, color='#004D40', edgecolors='black', linewidths=1,
                   s=0.107 * dot_scale, label='Mars')
        ax.annotate('♂', [1.54, 0.25], fontsize=12, usetex=False)

    return scatters

ani = FuncAnimation(
    fig, update, frames=len(cut_times), init_func=init,
    blit=False, interval=800, repeat=True
)

# Save if you want
#ani.save("paper_plots/cham_bi_tajer_animation.gif", dpi=200)
ani.save("paper_plots/cham_bi_tajer_animation.gif", dpi=200)

plt.show()
```
