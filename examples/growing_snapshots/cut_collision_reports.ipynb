```python
base_path = "/Users/honeyeah/Codes/data/" #folder with all data
n_runs = 130


sim_set = "ueda_v5/" #name of simulation set, folder with subfolders the name "sim_set_i"
sim_archive_path = base_path + sim_set + "ueda_v5_" #path given to get_runtime_data function
ueda_v5_rdf = rr.get_runtime_data(sim_archive_path, n_runs, "ueda_v5_") #rdf is runtime data frame

sim_set = "cham_bi_tajer/" #name of simulation set, folder with subfolders the name "sim_set_i"
sim_archive_path = base_path + sim_set + "cham_bi_tajer_" #path given to get_runtime_data function
cham_bi_tajer_rdf = rr.get_runtime_data(sim_archive_path, n_runs, "cham_simarchive_") #rdf is runtime data frame

sim_set = "cham_cont/" #name of simulation set, folder with subfolders the name "sim_set_i"
sim_archive_path = base_path + sim_set + "cham_cont_" #path given to get_runtime_data function
cham_cont_rdf = rr.get_runtime_data(sim_archive_path, n_runs, "cham_cont_") #rdf is runtime data frame


maxtimes_ueda_v5 = np.array([
    ueda_v5_rdf[ueda_v5_rdf['simulation_id'] == i + 1]['time'].max()
    for i in range(n_runs)
])

maxtimes_cham_bi_tajer = np.array([
    cham_bi_tajer_rdf[cham_bi_tajer_rdf['simulation_id'] == i + 1]['time'].max()
    for i in range(n_runs)
])

maxtimes_cham_cont = np.array([
    cham_cont_rdf[cham_cont_rdf['simulation_id'] == i + 1]['time'].max()
    for i in range(n_runs)
])

keep_ueda_v5 = np.where(maxtimes_ueda_v5 > 299.9e6)[0]
keep_cham_bi_tajer = np.where(maxtimes_cham_bi_tajer > 299.9e6)[0]
keep_cham_cont = np.where(maxtimes_cham_cont > 299.9e6)[0]

print("number of finished runs\n")
print(f"ueda_v5: {len(keep_ueda_v5)}")
print(f"cham_bi_tajer: {len(keep_cham_bi_tajer)}")
print(f"cham_cont: {len(keep_cham_cont)}")
```
